# Tensor ops

```python
my_torch = torch.arange(10)
my_torch
```

```python
# Reshape and view
my_torch = my_torch.reshape(2,5)
my_torch
```

```python
# Reshape if we don't know the number of items using -1
my_torch2 = torch.arange(10)
my_torch2
```

```python
my_torch2 = my_torch2.reshape(2, -1)
my_torch2
```

```python
my_torch3 = torch.arange(10)
my_torch3
```

```python
my_torch4 = my_torch3.view(2,5)
my_torch4
```

```python
# Update & mutation
my_torch5 = torch.arange(10)
my_torch5
```

```python
my_torch6 = my_torch5.reshape(2, 5)
my_torch6
```

```python
my_torch5[1] = 123
my_torch5
```

```python
my_torch6
```

```python
# Slices
my_torch7 = torch.arange(10)
my_torch7
```

```python
# Get an element
my_torch7[7]
```

```python
# get a slice
my_torch8 = my_torch7.reshape(5,2)
my_torch8
```

```python
my_torch8[:, 1]
```

```python
# return columns
my_torch8[:, 1:]
```
