# Tensor ops

```python
my_torch = torch.arange(10)
my_torch
```

```python
# Reshape and view
my_torch = my_torch.reshape(2,5)
my_torch
```

```python
# Reshape if we don't know the number of items using -1
my_torch2 = torch.arange(10)
my_torch2
```

```python
my_torch2 = my_torch2.reshape(2, -1)
my_torch2
```

```python
my_torch3 = torch.arange(10)
my_torch3
```

```python
my_torch4 = my_torch3.view(2,5)
my_torch4
```

```python
# Update & mutation
my_torch5 = torch.arange(10)
my_torch5
```

```python
my_torch6 = my_torch5.reshape(2, 5)
my_torch6
```

```python
my_torch5[1] = 123
my_torch5
```

```python
my_torch6
```

```python
# Slices
my_torch7 = torch.arange(10)
my_torch7
```

```python
# Get an element
my_torch7[7]
```

```python
# get a slice
my_torch8 = my_torch7.reshape(5,2)
my_torch8
```

```python
my_torch8[:, 1]
```

```python
# return columns
my_torch8[:, 1:]
```

# TENSOR MATH OPERATIONS

```python
tensor_a = torch.tensor([1, 2, 3, 4])
tensor_b = torch.tensor([5, 6, 7, 8])
```

```python
# Addition
tensor_a + tensor_b
```

```python
# Addition longhand
torch.add(tensor_a, tensor_b)
```

```python
# Subtraction
tensor_b - tensor_a
```

```python
# sub function
torch.sub(tensor_b, tensor_a)
```

```python
# multiplication
tensor_a * tensor_b
```

```python
# mul longhand
torch.mul(tensor_a, tensor_b)
```

```python
# division
tensor_b / tensor_a
```

```python
# div longhand
torch.div(tensor_b, tensor_a)
```

```python
# Remainder Modulus
tensor_b % tensor_a
```

```python
# Reminder longhand
torch.remainder(tensor_b, tensor_a)
```

```python
# Exponents / power
torch.pow(tensor_a, tensor_b)
```

```python
# Another way to write longhand
tensor_a.add(tensor_b)
```

```python
# Reassignment
tensor_a.add_(tensor_b)
```
